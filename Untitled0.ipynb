# MPG Data Analysis with Decision Tree and Random Forest Regressors

## 1. Import Libraries

```python
#import libraries
import seaborn as sns
import pandas as pd # Added for mpg_data.head() and y.head()
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
```

## 2. Load Data

```python
#load the data
mpg_data = sns.load_dataset('mpg')
display(mpg_data.head())
print(mpg_data.info())
#print (mpg_data.columns)
```

## 3. Data Preprocessing

```python
#drop rows with missing values
mpg_data = mpg_data.dropna()

#define the target and features
y = mpg_data['displacement']
features = ['mpg', 'cylinders', 'horsepower', 'weight', 'acceleration', 'model_year']
x = mpg_data[features]
```

## 4. Decision Tree Regressor (In-sample Prediction)

```python
#specifying the model
mpg_model = DecisionTreeRegressor(random_state=1)

#fit the model
mpg_model.fit(x, y)

#make predictions
predictions = mpg_model.predict(x)
print('First (in-sample) prediction:', predictions[:6])
print('Actual values:', y.head(6).tolist())
```

## 5. Data Splitting for Validation

```python
#split the data into training and validation data
train_x, val_x, train_y, val_y = train_test_split(x, y, random_state=1)
```

## 6. Decision Tree Regressor (Validation Prediction)

```python
#fit the model
mpg_model = DecisionTreeRegressor(random_state=1)
mpg_model.fit(train_x, train_y)

#make predictions on validation data
valid_predict = mpg_model.predict(val_x)

#print the first 6 validation predictions and actual values
print('Validation predictions:', valid_predict[:6])
print('Actual validation values:', val_y.head(6).tolist())

valid_error = mean_absolute_error(val_y, valid_predict)
print('Validation Mean Absolute Error:', valid_error)
```

## 7. Optimize Decision Tree Regressor (Finding Best `max_leaf_nodes`)

```python
#function to calculate mean absolute error for different max_leaf_nodes
def get_mae(max_leaf_nodes, train_x, val_x, train_y, val_y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=1)
    model.fit(train_x, train_y)
    preds_val = model.predict(val_x)
    mae = mean_absolute_error(val_y, preds_val)
    return(mae)

#compare different values of max_leaf_nodes
candidate_max_leaf_nodes = [5, 25, 50, 100, 250, 500]

#loop to find the ideal tree size from candidate_max_leaf_nodes
scores = {leaf_size: get_mae(leaf_size, train_x, val_x, train_y, val_y)
          for leaf_size in candidate_max_leaf_nodes}

#store the best value of max_leaf_nodes (it will be either 5, 25, 50, 100, 250 or 500)
best_tree_size = min(scores, key=scores.get)

#fill in argument to make optimal size and uncomment
final_model = DecisionTreeRegressor(max_leaf_nodes=best_tree_size, random_state=1)

#fit the final model and uncomment the next two lines
final_model.fit(x, y)
```

## 8. Random Forest Regressor

```python
#fit the random forest model
forest_model = RandomForestRegressor(random_state=1)
forest_model.fit(train_x, train_y)
forest_predictions = forest_model.predict(val_x)

#calculate the mean absolute error for the random forest model
forest_mae = mean_absolute_error(val_y, forest_predictions)
print('Random Forest Validation MAE:', forest_mae)
print('\n')
```
